```python
import re
def parse_charge(s):
  nums = re.findall(r'₹(\d+)', s)
  part = int(nums[0]) if nums else None
  labour = int(nums[1]) if len(nums) > 1 else None
  return pd.Series([part, labour])
df[['part_charge', 'labour_charge']] = df['subservice_charge'].apply(parse_charge)
df[['subservice_name', 'part_charge', 'labour_charge']].head()
```
